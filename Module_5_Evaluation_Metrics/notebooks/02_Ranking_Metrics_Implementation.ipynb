```python
# Hand-Calculated Verification Example# Recommended items: [10, 20, 30, 40, 50]# Known relevant ground truth: {20, 40, 99} (3 total relevant items)recommended = [10, 20, 30, 40, 50]relevant = {20, 40, 99}print("--- Metric Scores for Hand-Calculated Example ---")print(f"Precision@5: {precision_at_k(recommended, relevant, k=5):.4f} (Expected: 2/5 = 0.4000)")print(f"Recall@5:    {recall_at_k(recommended, relevant, k=5):.4f} (Expected: 2/3 = 0.6667)")print(f"F1@5:        {f1_at_k(recommended, relevant, k=5):.4f} (Expected: 0.5000)")print(f"DCG@5:       {dcg_at_k(recommended, relevant, k=5):.4f}")print(f"NDCG@5:      {ndcg_at_k(recommended, relevant, k=5):.4f}")print(f"AP@5:        {average_precision_at_k(recommended, relevant, k=5):.4f}")print(f"HitRate@5:   {hit_rate_at_k(recommended, relevant, k=5):.4f}")print(f"MRR@5:       {reciprocal_rank_at_k(recommended, relevant, k=5):.4f} (Hit at rank 2 -> 1/2 = 0.5)")
```
